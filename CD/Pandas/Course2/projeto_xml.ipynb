```python
import sqlalchemy

from sqlalchemy import create_engine, inspect, text
```

```python
engine = create_engine('sqlite:///:memory:')
```

```python
engine
```

```python
#Código omitido

url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'
```

```python
import pandas as pd
```

```python
dados = pd.read_csv(url)
```

```python
dados.head()
```

```python
dados.to_sql('clientes', con=engine, index=False)
```

```python
inspector = inspect(engine)
```

```python
print(inspector.get_table_names())
```

```python
query = 'SELECT * FROM clientes WHERE Categoria_de_renda = "Empregado"'
empregados = pd.read_sql(query, engine)
```

```python
empregados.head()
```

```python
empregados.to_sql('empregados', con=engine, index=False)
```

```python
pd.read_sql_table('empregados', con=engine)
```

```python
pd.read_sql_table('empregados', con=engine, columns=['ID_Cliente', 'Idade', 'Categoria_de_renda'])
```

```python
query = 'DELETE FROM empregados WHERE ID_Cliente=6840222'
with engine.connect() as connection:
    connection.execute(text(query))
    connection.commit()
```
