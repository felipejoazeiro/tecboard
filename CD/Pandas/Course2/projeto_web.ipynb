```python
import pandas as pd
```

```python
from pathlib import Path


# Procura o arquivo HTML em caminhos prováveis (cwd do notebook ou raiz do workspace)
candidates = [
    Path('content') / 'filmes_wikipedia.html',
    Path('CD') / 'Pandas' / 'Course2' / 'content' / 'filmes_wikipedia.html',
]


html_path = next((p for p in candidates if p.exists()), None)
if html_path is None:
    raise FileNotFoundError(
        "Não encontrei 'filmes_wikipedia.html'. "
        f"CWD={Path().resolve()} | tentei: {[str(p) for p in candidates]}"
    )


page = pd.read_html(html_path)
```
