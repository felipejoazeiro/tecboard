```python
import pandas as pd
```

```python
dados_pacientes = pd.read_json('./content/pacientes.json')
dados_pacientes.head()
```

```python
dados_pacientes_2 = pd.read_json('./content/pacientes_2.json')
dados_pacientes_2.head()
```

```python
pacientes = pd.json_normalize(dados_pacientes_2['Pacientes'])
pacientes.head()
```

```python
problemas_saude = pacientes['Problemas_saude'].explode().reset_index(drop=True)


problemas_saude = problemas_saude.astype(str).str.strip().replace({'diabetes': 'Diabetes'})


problemas_saude.head()
```

```python
pacientes.to_json('./content/pacientes_normalizados.json', orient='records', lines=True)
```
