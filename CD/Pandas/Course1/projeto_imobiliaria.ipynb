```python
imoveis_comerciais = ['Conjunto Comercial/Sala', 
                      'Prédio Inteiro', 'Loja/Salão', 
                      'Galpão/Depósito/Armazém', 
                      'Casa Comercial', 'Terreno Padrão',
                      'Loja Shopping/ Ct Comercial',
                      'Box/Garagem', 'Chácara',
                      'Loteamento/Condomínio', 'Sítio',
                      'Pousada/Chalé', 'Hotel', 'Indústria']
                                            
```

```python
dados.query('@imoveis_comerciais in Tipo')
```

```python
dados.query('Tipo not in @imoveis_comerciais')
```

```python
df = dados.query('@imoveis_comerciais not in Tipo')
```

```python
df.head()
```

```python
df.Tipo.unique()
```

```python
df_preco_tipo = df.groupby('Tipo')[['Valor']].mean().sort_values('Valor', ascending=True)
df_preco_tipo
df_preco_tipo.plot(kind='barh', figsize=(14, 10), color='purple', legend=False)
```

```python
df.Tipo.unique()
```

```python
df_percentual_tipo = df.Tipo.value_counts(normalize=True).to_frame().sort_values('Tipo')
```

```python
df_percentual_tipo.plot(kind='bar', figsize=(8, 6), color='orange', legend=False)
```
