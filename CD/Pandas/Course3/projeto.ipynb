```python
data.loc[:, 'Nível 1 - Setor' : 'Produto'].columns
```

```python
colunas_info = list(data.loc[:, 'Nível 1 - Setor' : 'Produto'].columns)
colunas_info
```

```python
data.loc[:, 1970:2021].columns
```

```python
colunas_emissao = list(data.loc[:, 1970:2021].columns)
colunas_emissao
```

```python
emissoes_por_ano = data.melt(id_vars=colunas_info, value_vars=colunas_emissao, var_name='Ano', value_name='Emissão')
emissoes_por_ano.head()
```

```python
emissoes_por_ano.groupby('Ano')
```

```python
emissoes_por_ano.groupby('Gás').groups
```

```python
emissoes_por_ano.groupby('Gás').get_group('CO2 (t)')
```

```python
emissao_total = emissoes_por_ano.groupby('Ano')['Emissão'].sum()
emissoes_por_ano.groupby('Gás')['Emissão'].sum()
```

```python
emissao_media = emissoes_por_ano.groupby('Ano')['Emissão'].mean()
emissao_media.plot()
```

```python
emissao_maxima = emissoes_por_ano.groupby('Ano')[['Emissão']].max()
emissao_maxima.plot()
```

```python
emissao_por_gas = emissoes_por_ano.groupby('Gás')[['Emissão']].sum().sort_values('Emissão', ascending=False)
emissao_por_gas
```

```python
emissao_por_gas.plot(kind='barh', figsize=(10, 6))
```

```python
emissao_por_gas.iloc[0:9]
```

```python
print(f'A emissão de CO2 é {emissao_por_gas.loc["CO2 (t)", "Emissão"]:.2f} toneladas, o que representa {emissao_por_gas.loc["CO2 (t)", "Emissão"] / emissao_por_gas["Emissão"].sum() * 100:.2f}% do total de emissões.')
```
